# Generator Layer Test
Tests batch generation and quota scheduling across topic x difficulty buckets.

```python
from datasetgen.generator import BatchGenerator, QuotaScheduler
from datasetgen.pipeline import DatasetPipeline
from datasetgen.providers import get_provider

provider = get_provider('mock')
pipeline = DatasetPipeline(provider)
blueprint = pipeline.plan('I need 50 mechanics physics questions')

scheduler = QuotaScheduler(blueprint)
generator = BatchGenerator(provider)

topic, diff, needed = scheduler.get_next_target_bucket()
examples, resp = generator.generate_batch(blueprint, topic, diff, count=5, batch_index=0)
print(f'Generated {len(examples)} examples for {topic}:{diff}.')
print('Sample question:', examples[0].data['question'])
```
