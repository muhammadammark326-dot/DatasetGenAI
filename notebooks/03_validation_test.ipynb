# Validation Layer Test
Validates examples programmatically with SymPy and schema checkers.

```python
from datasetgen.validators.registry import ValidatorRegistry
from datasetgen.pipeline import DatasetPipeline
from datasetgen.providers import get_provider

provider = get_provider('mock')
pipeline = DatasetPipeline(provider)
blueprint = pipeline.plan('I need 20 high-school physics questions')

chain = ValidatorRegistry.build_chain(blueprint)
print('Active validator chain:', [v.identifier for v in chain])

# Test correct item
sample_ok = {
    'question': 'A block with mass 5 kg accelerates across a frictionless surface at 10 m/s^2. What is the net force?',
    'answer': '50.0 N',
    'explanation': 'F = m * a = 5 * 10 = 50 N',
    'topic': 'mechanics',
    'difficulty': 'beginner'
}
for v in chain:
    res = v.validate(sample_ok, blueprint)
    print(f'{v.identifier}: {res.status}')
```
